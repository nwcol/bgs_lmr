```python
import numpy as np
```

```python
# Load YRI sample IDs
sample_ids = []
with open("YRI_sample_ids.txt") as fin:
    for line in fin:
        sample_ids.append(line.rstrip("\n"))

# Path to 
PATH = "/media/nick/Elements/Data/NYGC_GVCFs/allele_count_arrays/"
suffix = lambda chrom: f"_allele_counts_chr{chrom}.npy"

for chrom in range(1, 23):
    summed_counts = 0
    for sample_id in sample_ids:
        sample_counts = np.load(PATH + sample_id + suffix(chrom))
        summed_counts += sample_counts
        print(chrom, sample_id)
    np.save(f"YRI/nrna_YRI_chr{chrom}.npy", summed_counts)
```
