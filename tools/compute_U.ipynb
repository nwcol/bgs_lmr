```python
import bgshr
import numpy as np
import pandas
```

```python
ccds_lof_deciles = [f"cds_lof_d{i}" for i in range(1, 12)]
phastcons_bins = [f"phastcons_{(5*i)}-{5*(i+1)}" for i in range(12)]
elem_classes = ["cds_merged", "promoters", "enhancers"] + ccds_lof_deciles + phastcons_bins
chroms = list(range(1, 23))
mut_models = ["carlson", "gnomad", "roulette"]
```

```python
# Tabulate total deleterious mutation rates

cols = ["chrom", "mut_model", "elem_type", "U"]
data = []

for chrom in chroms:
    for mut_model in mut_models:
        mut_fname = f"../data/mutation_arrays/{mut_model}/{mut_model}_map_chr{chrom}.npy"
        mut_map = np.load(mut_fname)
        mut_map = np.ma.array(mut_map, mask=np.isnan(mut_map))

        for elem_class in elem_classes:
            # Different element types have slightly different filepaths
            # Merged CCDS
            if elem_class == "cds_merged":
                annot_fname = f"../data/annotations/cds/merged/cds_merged_chr{chrom}.bed.gz"

            # CCDS LOF deciles
            elif elem_class in ccds_lof_deciles:
                x = elem_class.replace("cds_", "")
                annot_fname = f"../data/annotations/cds/{x}/{elem_class}_chr{chrom}.bed.gz"

            # Regulatory elements
            elif elem_class in ["promoters", "enhancers"]:
                annot_fname = f"../data/annotations/regulatory/{elem_class}/{elem_class}_chr{chrom}.bed.gz"

            # Non-coding phastCons elements
            elif elem_class in phastcons_bins:
                x = elem_class.replace("phastcons_", "")
                annot_fname = f"../data/annotations/phastcons/top{x}/phastcons_top{x}_chr{chrom}.bed.gz"

            else:
                raise ValueError("unknown element class")

            elements = bgshr.Util.read_bedfile(annot_fname)
            mask = bgshr.Util.elements_to_mask(elements, L=len(mut_map))
            n_sites = np.sum(mask == False)
            elem_mut_map = np.ma.array(mut_map, mask=mask)
            n_sites_masked = np.sum(elem_mut_map.mask == False)
            U = np.sum(elem_mut_map) * n_sites / n_sites_masked
            row = [
                chrom, 
                mut_model, 
                elem_class, 
                U
            ]
            data.append(row)
            print(chrom, mut_model, elem_class)

df = pandas.DataFrame(data, columns=cols)
df.to_csv("../data/U_tbl.csv", index=False)
```
