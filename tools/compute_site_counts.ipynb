```python
mut_models = [
    "roulette",
    "gnomad",
    "carlson"]
chroms = list(range(1, 23))

cols = [
    "mut_model",
    "chrom",
    "num_sites"]
data = []

for mut_model in mut_models:
    for chrom in chroms:
        fname = f"{prefix}/strict_{mut_model}/mask_chr{chrom}.bed.gz"
        regions = bgshr.Util.read_bedfile(fname)
        num_sites = np.sum(regions[:, 1] - regions[:, 0])
        row = [
            mut_model,
            chrom,
            num_sites]
        data.append(row)
df = pandas.DataFrame(data, columns=cols)
df.to_csv("../data/site_count_tbl.csv", index=False)
```
