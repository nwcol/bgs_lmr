```python
import numpy as np
import bgshr
import pandas
import scipy
import matplotlib.pyplot as plt
```

```python
def apply_threshold(df, thresh):
    # drop all windows where df["num_sites"] < thresh
    df = df[df["num_sites"] >= thresh]
    return df
```

```python
demographies = [
    "equilibrium_granular_Ne"]
cons_models = [
    "merged_cds_regulatory",
    "split_cds_regulatory",
    "merged_cds_phastcons",
    "split_cds_phastcons"]
mut_models = [
    "carlson",
    "gnomad",
    "roulette"]
chroms = list(range(1, 23))
scales = [1e4, 1e5, 1e6]
threshs = [7_500, 50_000, 250_000]

cols = [
    "cons_model",
    "mut_model",
    "chrom",
    "R^2_B,pi_10kb",
    "R^2_Epi,pi_10kb",
    "R^2_B,pi_100kb",
    "R^2_Epi,pi_100kb",
    "R^2_B,pi_1Mb",
    "R^2_Epi,pi_1Mb",
]
data = []

cols_to_keep = [
    "chrom", 
    "chromStart", 
    "chromEnd",
    "num_sites",
    "B",
    "exp_pi",
    "avg_pi"
]

for demography in demographies:
    for cons_model in cons_models:
        for mut_model in mut_models:
            for chrom in chroms:
                fname = f"../models/tables_10kb/{demography}/{cons_model}/{mut_model}/B_tbl_YRI_chr{chrom}_10kb.csv.gz"
                df = pandas.read_csv(fname)
                reduced_df = df.loc[:, cols_to_keep]
                row = [
                    cons_model,
                    mut_model,
                    chrom]
                for scale, thresh in zip(scales, threshs):
                    scaled_df = bgshr.Util.scale_genome_table(df, scale)
                    scaled_df = apply_threshold(scaled_df, thresh)
                    row += [
                        scipy.stats.pearsonr(scaled_df["B"], scaled_df["avg_pi"]).statistic ** 2,
                        scipy.stats.pearsonr(scaled_df["exp_pi"], scaled_df["avg_pi"]).statistic ** 2]
                data.append(row)
                print(cons_model, mut_model, chrom)
df = pandas.DataFrame(data, columns=cols)
df.to_csv("../models/R2_tbl.csv", index=False)
```
